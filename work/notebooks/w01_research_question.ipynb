## The question: decision, action, cost of a wrong call

The decision this work improves is deciding which content pages should be reviewed first when a team has limited time and resources.

The person acting on this recommendation is a content or SEO team member who can decide whether a page should be refreshed, expanded, monitored, or investigated further.

A wrong recommendation has two costs. First, the team may spend time improving pages that do not have meaningful opportunities. Second, important pages with performance issues may be missed and remain unresolved.

This project aims to support better prioritization using measurable signals from the data, not to automatically decide what changes will guarantee better performance.

## Quick look at the data

The starter dataset contains 30,000 anonymized content pages, which makes manual review difficult and creates a need for prioritization.

A first look at the data shows that the average page received 5,200.37 impressions and 16.1 clicks over the last 90 days. This suggests that pages can receive visibility while having very different levels of engagement.

Additionally, 16,262 pages show a declining trend direction. This indicates that there may be a useful opportunity to identify pages that deserve review first based on measurable signals.

These numbers do not prove that a page will improve after changes, but they suggest that a decision-support system for prioritizing reviews is worth exploring.


```python
import pandas as pd

df = pd.read_csv(
    "flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"
)

print(df.shape)
```

```python

# Dataset size
print("Total pages:", len(df))

# Average search visibility and engagement signals
print("Average impressions (90d):", round(df["impressions_90d"].mean(), 2))
print("Average clicks (90d):", round(df["clicks_90d"].mean(), 2))

# Pages showing declining trend
print(
    "Pages with declining trend:",
    (df["trend_direction"] == "down").sum()
)

# Pages not updated recently
print(
    "Pages with more than 180 days since update:",
    (df["days_since_last_update"] > 180).sum()
)
```
